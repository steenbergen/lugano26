## 4.3 Core Data Types & Data Type Conversions
Python has three main scalar families: numbers (int/float), strings, and  Booleans. Converting ("casting") between them is a daily task when preparing  data for models -- e.g., CSV files arrive as text, but models need numbers.  The casting functions are str(), int(), and float().

```python
# Number -> String. You cannot concatenate a number onto a string with `+`
# directly (Python raises a TypeError), so cast it with str() first.
price = 299.99
print("The Price Is " + str(price))

# String -> Integer. int() parses the text "1998" into the number 1998,
# after which normal arithmetic works.
birth_year_str = "1998"
birth_year = int(birth_year_str)
age = 2026 - birth_year
print(f"Age: {age}")

# String -> Float. Use float() when the text contains a decimal point --
# int("98.5") would fail, but float("98.5") works.
score = "98.5"
numeric_score = float(score)
print(numeric_score * 2)
```

# 5. Data Structures and Manipulation
## 5.1 Indexing and Slicing
Accessing elements by 0-based position is critical in deep learning, where you constantly slice data tensors and target sequences. The same [start:stop] syntax you see here on a string works identically on lists, NumPy arrays, and PyTorch tensors.

Rules to remember:


*   Indexing starts at 0, so text[0] is the FIRST character.
*   Negative indices count from the end: text[-1] is the LAST character.
*   A slice [start:stop] INCLUDES start but EXCLUDES stop ("half-open interval"). For instance, `text[0:4]` gives characters 0, 1, 2, 3.
*   Omitting a bound means "to the end" (or "from the start"): `text[5:]`     runs from index 5 through the last character.


```python
text = "Deep Learning"
print("First char:", text[0])        # 'D'
print("Last char:", text[-1])        # 'g'
print("Slice (0:4):", text[0:4])     # 'Deep'  (indices 0-3)
print("Slice (5 onwards):", text[5:])  # 'Learning'
```

## 5.2 Lists
Lists are ordered, MUTABLE (changeable) collections, written with square brackets []. In machine learning they hold data sequences, loss histories  recorded during training, lists of model layers, and much more. Of all the data structures, these and dictionaries are the most widely used in machine learning.

There are two ways to grow a list:


1.   `append(x)` adds ONE element to the end.
2.   `extend([a, b])` adds EVERY element of another list to the end.

```python
losses = [0.85, 0.62, 0.41]     # e.g., loss after epochs 1-3
losses.append(0.28)             # one more epoch
losses.extend([0.19, 0.12])     # two more epochs at once
print("Loss history:", losses)
```

## 4.3 List Comprehension and Filtering
Among the most important Pythonic operations is list comprehension. It builds a new list from an existing sequence in a single readable line. The general pattern is:

[ <expression> for <item> in <sequence> if <condition> ]

This replaces a 3-4 line for-loop with append() calls, and is the idiomatic way to transform and filter data. R users: think of it as sapply()/purrr::map() combined with a filter.

```python
# Transforming: square every number from 0 to 9.
# range(10) generates 0, 1, ..., 9 (stop value excluded, like slicing).
squares = [x**2 for x in range(10)]
print("Squares:", squares)
```

```python
# Filtering: keep only even numbers. The `if` clause acts as the filter;
# % is the modulus operator (remainder after division), so x % 2 == 0
# is true exactly when x is even.
evens = [x for x in range(10) if x % 2 == 0]
print("Evens:", evens)
```

```python
# A realistic data-cleaning example: raw class labels often carry stray
# whitespace and inconsistent case. Chaining string methods inside a
# comprehension normalizes the whole list at once:
#   .strip() removes leading/trailing whitespace
#   .upper() converts to uppercase
labels = [" cat ", " dog ", " bird "]
clean_labels = [label.strip().upper() for label in labels]
print("Cleaned Labels:", clean_labels)
```

## 4.4 Tuples, Sets, and Dictionaries
Beyond lists, Python has three other core containers:


1.   **Tuples ()**: IMMUTABLE (fixed) ordered sequences. Once created they      cannot be changed, which makes them ideal for things that should never be accidentally modified, such as array/tensor shapes like (batch_size, channels, height, width).
2.   ** Sets {}**: UNORDERED collections of UNIQUE elements. Duplicates are    removed automatically, which makes sets perfect for finding the unique      class labels in a dataset.
3.   **Dictionaries {key: value}**: key-value pairs, like a lookup table.     Essential for hyperparameter settings, model configurations, and JSON-like data. (Note: sets and dicts both use curly braces; a dict always has colons between keys and values.)

```python
# Tuple "unpacking" assigns each element to its own variable in one step --
# a very common idiom for pulling apart tensor shapes.
input_shape = (32, 3, 224, 224)          # (batch, channels, height, width)
batch_size, channels, h, w = input_shape
print(f"Batch size: {batch_size}, Channels: {channels}")
```

```python
# Passing a list to set() deduplicates it. Note the result is unordered:
# the printout order may differ from the input order (and between runs).
raw_classes = ["cat", "dog", "cat", "bird", "dog"]
unique_classes = set(raw_classes)
print("Unique classes:", unique_classes)
```

```python
# A dictionary of training hyperparameters -- you will see configs exactly
# like this throughout PyTorch code. Values are looked up by key, e.g.
# config["learning_rate"].
config = {
    "learning_rate": 0.001,
    "epochs": 10,
    "optimizer": "Adam"
}
print(config)
```

```python
# .items() yields (key, value) pairs, which the for-loop unpacks into two
# loop variables -- the standard way to iterate over a dictionary.
for key, value in config.items():
    print(f"{key}: {value}")
```

## 7.1 Conditional Checks
The branches are tested top to bottom and the FIRST true condition wins -- here 0.05 < 0.1, so only the first branch runs.

```python
loss = 0.05
if loss < 0.1:
  print("Model converged well!")
elif loss < 0.5:
  print("Model is training...")
else:
  print("Loss too high: check learning rate")
```

# 8 Functions
Functions are reusable, named blocks of code, defined with `def`. Wrapping operations in functions keeps ML pipelines clean, testable, and free of  copy-paste duplication.

Anatomy of the function below:


*   `def calculate_accuracy(predictions, targets):`  -- name + parameters.
*   The triple-quoted string is a "docstring" -- documentation that lives     inside the function and is shown by help(calculate_accuracy).
*   `zip(a, b)` pairs the two lists element-by-element, so (p, t) walks     through matched prediction/target pairs.
*   `p == t` is True/False, and Python treats True as 1 and False as 0, so sum() of those comparisons counts the correct predictions.
*   `return` sends the result back to the caller.

```python
def calculate_accuracy(predictions, targets):
  """Calculates simple percentage accuracy."""
  correct = sum([p == t for p, t in zip(predictions, targets)])
  return correct / len(targets)

preds = [1, 0, 1, 1]
true_y = [1, 0, 0, 1]
acc = calculate_accuracy(preds, true_y)
print(f"Accuracy: {acc * 100:.1f}%")
```

# 11 Object-Oriented Programming
Up until now, we have been writing code as a list of step-by-step instructions. Object-Oriented Programming (OOP) is a shift in mindset: instead of just thinking about actions, we start building digital "things" that combine both data and behavior into a single package—much like designing a blueprint before building real-world objects.

Python shines at OOP and that is a good thing, since it plays a major role in deep learning. Ecosystems such as PyTorch rely on classes. Those ecosystems, in turn, are the key to modern deep learning.

## 11.1 What Is a Class?
Most code you have written is probably a sequence of function calls on data that sits in separate variables. A class bundles the two together: it is a blueprint that packages data (called attributes) with the functions that operate on that data (called methods). From one blueprint you can stamp out many concrete copies, called instances or objects.

You have been using classes all along without noticing. When you write `first_name.upper()`, `first_name` is an instance of the built-in class `string` and `upper()` is a method we apply to that instance. We are no limited to built-in classes, however. It is easy enough to start writing some of our own.

## 11.2 Building a Class from Scratch
Imagine we teach a course with students (attribute) who obtain grades (another attribute) throughout the semester. At the end of the semester, we would like to compute the average grade. The essential attribute is the name of the student. We need this even if students never hand in a single assignment after they enroll. The following code builds the class.


```python
class Student:
    def __init__(self, name):      # runs once, when the instance is created
        self.name = name           # store data ON the instance
        self.grades = []           # a list of grades

    def add_grade(self, g):        # a method: a function attached to the instance
        self.grades.append(g)

    def average(self):             # another method
        return sum(self.grades) / len(self.grades)
```

There are three things to internalize:


*   `__init__` is the constructor: it runs once when the instance is created and sets up its data.
*   `self` is how the object refers to itself. Anything stored as `self.something` in `__init__` is available in every method later. (Python passes self automatically---you never supply it when calling.)
*   **Inheritance**: class Student(Person) would mean: Student gets everything Person has, plus whatever it adds. The class in parentheses is the parent.

## 11.3 Building Instances
Instances are built starting with the parent--other attributes can then be added later.

```python
alice = Student("Alice")           # calling the class creates an instance (runs __init__)
alice.add_grade(5.5)               # Midterm grade
alice.add_grade(6.0)               # Final exam grade
print(alice.average())             # 5.75
```

# 12.1 The Lambda Function
Earlier, we saw how to define a function using `def`. We use this command to define a function we plan on using multiple times. The function name becomes part of the **namespace**, ready to be called whenever we need it. However, we sometimes only want to call a function once--right where we need it--and we do not want to polute the namespace. In that case, we can define a **lambda function**. The most common you will see lambdas used as intended is inside functions like `sored()`, `map()`, or `filter`.

Imagine we keep a list of students and their grades. (Notice the use of tuples--we want the names and grades to be immutable.) We want to sort students by their grades. The roundabout way of achieving this goal is to write a whole function for this.

```python
# A list of student tuples: (Name, Score)
students = [("Alice", 88), ("Bob", 95), ("Charlie", 78)]

# Creating a full function just to extract the score for sorting once
def get_score(student):
    return student[1]

sorted_students = sorted(students, key=get_score)
print(sorted_students
```

A much easier way is to use a lambda function.

```python
# Same result, but inline and disposable
sorted_students = sorted(students, key=lambda student: student[1])
print(sorted_students)
```

The anatomy of a lamnda function is the following:


*   Keyword: lambda
*   Parameters: x, y, etc.
*   Colon
*   Expression: returned value

For the earlier example, `key = lambda`, parameter = `student`, the colon is in place, and expression = `student[1]`. `sorted()` uses this returned number to perform the comparison between students.

## 12.2 Flexible Arguments
You will run into `args` and `kwargs` in Python code. When I first did, I had no idea what they were. Over time, I learnt it is best to think of them as packing boxes for extra arguments.

Normally, Python functions are strict. If a function expects 2 arguments, you must pass 2 arguments. `*args` and `**kwargs` are Python's way of saying: "I don't know how many inputs the user will pass, so just pack whatever they send into a single variable." This can be extreely useful with unstructured data.

`*args` and `**kwargs` serve different functions:


*   `*args` packs positional inputs into a Tuple (a simple list-like package).
*   `**kwargs` packs named/keyword inputs into a Dictionary (key-value pairs).

*Note:* The names `args` and `kwargs` are just standard conventions! The real magic is the asterisks (* and **). You could technically write `*stuff` or `**options`, but sticking to `*args` and `**kwargs` keeps your code readable to other Python developers.

Below are two examples.

```python
# Example with *args
def calculate_total(*args):
    # 'args' inside the function is just a plain tuple: (10, 20, 5)
    print("The packed args tuple:", args)

    total = sum(args)
    return total

print(calculate_total(10, 20, 5))
# Output tuple: (10, 20, 5)
# Returns: 35
```

```python
# Example with **kwargs
def build_profile(username, **kwargs):
    # 'kwargs' inside the function is a plain dictionary!
    print("The packed kwargs dict:", kwargs)
    print(f"User: {username}")

    for key, value in kwargs.items():
        print(f"  {key}: {value}")

build_profile("alex99", age=25, city="Chicago", status="Active")
# Output dict: {'age': 25, 'city': 'Chicago', 'status': 'Active'}
```
